# Wearable Prototype Hardware Overview

## Core Components
- **ESP32 DevKit V1**: Main controller; reads all sensors, processes data, sends it to Firebase Cloud
- **MAX30102 Pulse Oximeter Module**: Measures raw heart rate and SpO₂ signals using IR & RED LEDs
- **MPX5010 / MPX5050 Pressure Sensor**: Reads raw blood pressure signal from cuff or pressure source; later software converts to BP values
- **16x16 LED Matrix**: Displays visual alerts or simple numeric/graphic feedback for heart rate, SpO₂, or BP; highly visible even from a distance
- **Active Buzzer**: Generates sound alerts if vitals cross predefined thresholds


## Power Components
- **Li-ion / Li-Po Battery (3.7V)**: Portable power supply for wearable setup
- **TP4056 Charging Module**: Allows safe recharging of Li-ion battery
- **Slide Switch (Optional)**: Turns wearable device ON/OFF easily

## Prototyping / Integration Tools
- **Breadboard**: Temporary platform for assembling and testing components
- **Jumper Wires (Male–Male / Male–Female)**: Connects sensors, LED matrix, and ESP32 for prototyping
- **Resistors / Capacitors**: Optional; for sensor signal stability and noise reduction
- **Enclosure / Wristband**: Makes the device wearable and user-friendly
