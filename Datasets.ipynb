# Dummy Dataset for BAN Project

## Dataset URL
https://data.mendeley.com/datasets/xdrwrh9zbk/2

## How to Use This Dataset for Raw Processing

### Heart Rate (HR)
- **Use either IR or Red channel** from PPG
- **Steps:**
  1. Treat the column as a waveform over time
  2. Detect peaks → each peak = one heartbeat
  3. Calculate intervals between peaks → BPM

### SpO₂
- **Use both Red and IR channels**
- **Steps:**
  1. Extract AC (pulsatile) and DC (baseline) components from each channel
  2. Compute **ratio of ratios**:  

     \[
     R = \frac{AC_{red}/DC_{red}}{AC_{IR}/DC_{IR}}
     \]

  3. Apply calibration formula → SpO₂ %

### Blood Pressure (BP)
- Cannot be extracted from Red/IR PPG alone
- For BP estimation, you would need:
  - Pressure sensor waveform (e.g., MPX5010 / MPX5050), or
  - ECG + PPG for **Pulse Transit Time (PTT)** method

## Why This Dataset Is Useful
- Provides **raw waveform data** (like what a MAX30102 sensor outputs)
- Can **simulate HR & SpO₂ calculation pipeline** before connecting actual hardware
- Optional metadata (Age, Gender, Hemoglobin) can assist **calibration or analysis**



## Dataset URL
https://archive.ics.uci.edu/dataset/340/cuff+less+blood+pressure+estimation

## Dataset Type
- **Format:** MATLAB v7.3 `.mat` file → can be opened in MATLAB or Python (`h5py`)
- **Structure:** cell array of matrices → each cell = one recording session  
  - Each matrix: **rows = signal channels**, **columns = samples over time**

## Signal Channels
| Row | Signal          | Sampling Frequency | Notes                                                                     |
| --- | --------------- | ----------------- | ------------------------------------------------------------------------- |
| 1   | PPG (fingertip) | 125 Hz            | Raw waveform → can calculate **HR & SpO₂**                                |
| 2   | ABP             | 125 Hz            | Arterial blood pressure waveform (mmHg) → ground truth for BP validation  |
| 3   | ECG             | 125 Hz            | Channel II ECG → use for **pulse transit time (PTT)** or HR               |

## Why It’s Useful
- **PPG:** raw waveform → simulates MAX30102 output
- **ECG:** raw waveform → needed for **BP via PTT**
- **ABP:** raw waveform → ground truth for BP estimation
- **Sampling frequency 125 Hz** → sufficient for HR, SpO₂, and BP processing

## How to Use It
1. **Load `.mat` file** (MATLAB or Python `h5py`)
2. **Extract channels:**
   - `PPG` → calculate **HR & SpO₂**
   - `ECG` → detect R-peaks for **PTT**
   - `ABP` → validate **estimated BP**
3. **Process waveform:**
   - Peak detection → **HR**
   - AC/DC ratio → **SpO₂**
   - PTT (ECG → PPG delay) → **BP estimation**

## Summary
- Raw, unprocessed waveform dataset
- Contains **PPG, ECG, and ABP** → all vital signs needed for BAN project
- Ideal for **simulating HR, SpO₂, and BP calculations** from hardware like MAX30102 + MPX5010
