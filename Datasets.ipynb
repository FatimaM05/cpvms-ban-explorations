# Dummy Datasets for Initial Working

## https://www.physionet.org/content/pulse-transit-time-ppg/1.1.0/#files-panel

- This dataset contains **raw multi-sensor physiological signals** similar to what our hardware will capture.
- **PPG channels (pleth_1 to pleth_6, red/IR/green wavelengths)** from finger sensors → can be processed to calculate **SpO₂** and **pulse-based heart rate**.
- **ECG signals and annotated R-peaks** → help validate and refine **heart rate calculation**.
- **Systolic and diastolic blood pressure columns** → allow estimation of **blood pressure** using **pulse transit time (PTT)**.
- **Load cell (sensor pressure) and IMU signals** → can be used to handle **motion artifacts** and **sensor contact quality**.
- **Temperature channels (temp_1, temp_2, temp_3)** → can be used later for **temperature analysis**.
- Data is **time-synchronized and high resolution** → suitable for generating **dummy/raw values** in Python/Jupyter, simulating what the hardware will capture.
- By processing these columns, we can derive **SpO₂, heart rate, and BP**, just as we will do with our BAN sensors in real-time.

## https://www.physionet.org/content/wearable-exam-stress/1.0.0/data/S1/Final/#files-panel

A Wearable Exam Stress Dataset for Predicting Cognitive Performance in Real-World Settings

- **Ignore** `HR.csv` and `IBI.csv` (already processed)
- **Take** `BVP.csv` → process peaks for **Heart Rate**, AC/DC for **SpO₂** (actually main raw source, equivalent to ADC counts coming from a hardware sensor)
- **Use** `ACC.csv` optionally to remove **motion artifacts**
- **Keep** `TEMP.csv` for **temperature** if you need environmental/skin data

#### How to Use This Dataset for Raw Processing

* Heart Rate (HR)
- **Do NOT use `HR.csv`** if you want to simulate processing like hardware
- **Use `BVP.csv`** → raw PPG waveform
- **Steps:**
  1. Load waveform
  2. Detect peaks → calculate RR intervals → BPM

* SpO₂
- **Use `BVP.csv`**
- If multiple channels (red & IR) are available → calculate **AC/DC ratio → SpO₂**
