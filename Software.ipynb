# Software Flow

## 1 Overall System Overview (High-Level)

### Data Source

* Patient vitals (SpO2, Heart Rate, BP) coming from sensors → device → server

### Backend / Server

* Stores patient data
* Runs alert logic
* Runs predictive analysis
* Runs Sepsis Detection and Length of Stay (LoS) / Discharge Readiness modules
* Serves data to dashboards

### Frontend Applications

* Parent Portal (limited access, single patient, simplified predictive messages)
* Staff Dashboard (multi-patient, real-time monitoring, actionable insights)

---

## 2 User Roles & Access Control

### Role 1: Parent / Guardian

* Logs in using Patient ID
* Can see only their patient
* Read-only access (no editing)
* Views simplified recovery/progress messages from LoS module

### Role 2: Hospital Staff

* Logs in using Hospital ID + Password
* Can see all patients
* Real-time monitoring
* Alert supervision
* Access to Sepsis Risk and Discharge Readiness % for each patient

Role-based access is handled after login by backend permissions.

---

## 3 Software Flow (Step-by-Step)

### Login Flow (Common Entry)

1. User opens application
2. Login screen appears
3. User selects role: Parent or Staff
4. Credentials are sent to backend
5. Backend authenticates user and identifies role
6. User is redirected to role-specific dashboard

---

## 4 Parent Portal – Software Flow

### Parent Dashboard Flow

Purpose: Simple, calm, informative

After Login:

1. Backend verifies Parent ↔ Patient mapping
2. Parent Dashboard loads

Dashboard Shows:

* Patient basic info
* Current vitals: SpO2, Heart Rate, Blood Pressure
* Status indicator: Normal / Warning / Danger
* Simplified predictive messages: Recovery Progress, Vitals stable or Risk detected – staff notified

24-Hour Data:

* Graphs for SpO2, HR, BP
* Averages: Avg SpO2, Avg HR, Avg BP

Parents do not see raw predictive scores or Sepsis analysis — only simple, reassuring messages.

---

## 5 Staff Dashboard – Software Flow

### Main Staff Dashboard (Real-Time View)

Purpose: Fast decision-making, replicates real hospital workflow

After Login:

1. Staff Dashboard loads
2. Backend streams real-time data

Dashboard Layout:

* Patient tiles/cards (visual triage)
* Each tile shows: Patient ID, SpO2, HR, BP, Status color (Normal / Alert / Danger)
* Module Indicators: Sepsis Risk Badge, Discharge Readiness %

Real-World Behavior:

* Staff scan tiles first, click only yellow/red tiles
* Tiles refresh automatically
* Immediate alerts displayed for critical vitals or predictive risks

Alert Logic:

* Backend continuously checks thresholds, Sepsis module, LoS module
* If exceeded: Tile turns yellow/red, alert is raised, predictive badge updated

---

### Clicking a Patient Tile

1. Staff clicks on Patient ID
2. Detailed Patient View opens

Detailed View Shows:

* Real-time vitals (live updating)
* 24-hour history graphs
* Averages & trends
* Alert history
* Predictive risk indicator (Low / Medium / High)
* Sepsis Risk Details
* Discharge Readiness Progress (%)

Staff can make fast, actionable decisions mimicking ICU/ward workflow.

---

### Screens in Patient Detail

Screen 1: Real-Time Vitals

* Shows live updating: SpO2, Heart Rate, Blood Pressure
* Status color indicators (Normal / Warning / Danger)

Screen 2: 24-Hour History / Trends

* Graphs for SpO2, HR, BP
* Includes averages and trend lines

Screen 3: Alert History

* List of past alerts for this patient
* Timestamped entries with vital values at alert time
* Can filter by Critical alerts or Warnings

Screen 4: Predictive Analysis

* Shows Risk indicator: Low / Medium / High
* Sepsis Risk probability
* Discharge Readiness Progress (%)
* Visual badges / charts for fast decision-making

Optional: Notes / Staff Actions Screen

* Staff can add or view Interventions, Observations, Recommendations

---

## 6 Alert & Status Logic (Backend Controlled)

Status Rules Example:

* SpO2: Normal ≥95%, Warning 90–94%, Danger <90%
* HR: Normal 60–100, Warning slight variation, Danger critical
* BP: Normal range, Warning elevated, Danger high/low

Backend:

* Evaluates raw data
* Assigns status & triggers alerts
* Runs Sepsis & LoS predictive modules
* Sends all information to frontend

Frontend:

* Displays colors, alert icons, predictive badges
* Parents see simplified messages only

---

## 7 Predictive Analysis

Prediction Targets:

* SpO2
* Heart Rate
* Blood Pressure

Recommended Predictions:

1. Risk Prediction: Predict risk of deterioration in next X hours (Low / Medium / High)
2. Trend Prediction: Detect downward trends for SpO2, instability for HR
3. Sepsis Prediction: Detect potential infection using vitals triad, output risk probability, triggers alert badge
4. Length of Stay / Discharge Readiness: Monitor stability over 6–24 hours, predict recovery phase (<50% Acute, 50–80% Stabilizing, 90% Ready for discharge)

Flow:

* Backend collects historical + live data
* Runs predictive modules
* Stores prediction and triggers alerts
* Frontend displays: Staff → detailed badges, charts; Parents → simplified status messages
